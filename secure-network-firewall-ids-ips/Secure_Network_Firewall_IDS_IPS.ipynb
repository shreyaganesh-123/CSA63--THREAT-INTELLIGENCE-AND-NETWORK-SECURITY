```python
# Firewall Rules + IDS/IPS Security Demonstration

firewall_rules = [
    ("Internet", "Web-Server", "TCP/80", "ALLOW"),
    ("Internet", "Web-Server", "TCP/443", "ALLOW"),
    ("Internet", "Database-Server", "TCP/3306", "DENY"),
    ("User-Network", "Database-Server", "TCP/3306", "ALLOW"),
    ("Unknown", "Internal-Network", "ANY", "DENY")
]

print("=" * 70)
print("          FIREWALL RULE AND IDS/IPS DEMONSTRATION")
print("=" * 70)

print("\nFIREWALL RULES")
print("-" * 70)

for i, rule in enumerate(firewall_rules, 1):
    source, destination, service, action = rule
    print(f"Rule {i}: {source} -> {destination} | {service} | {action}")

print("\nIDS/IPS DETECTION")
print("-" * 70)

events = [
    ("203.0.113.50", "Web-Server", "Port Scan", "HIGH", "BLOCK"),
    ("203.0.113.51", "Internal Service", "Repeated Login Failures", "HIGH", "BLOCK"),
    ("203.0.113.52", "Web-Server", "Excessive Connections", "CRITICAL", "BLOCK")
]

for source, target, activity, severity, action in events:
    print(f"ALERT: {activity}")
    print(f"Source: {source}")
    print(f"Target: {target}")
    print(f"Severity: {severity}")
    print(f"Action: {action}")
    print("-" * 70)
```
